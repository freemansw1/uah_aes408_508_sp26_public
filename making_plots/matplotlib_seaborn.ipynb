```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
ky_dams = pd.read_csv("../example_data/Ky_Water_Resources_Regulated_Dams.csv")
```

```python
fig, axes = plt.subplots(1,2)

ax1 = axes[0]
ax2 = axes[1]

sns.histplot(
    ky_dams, x="LENGTH", ax=ax1
)
sns.histplot(
    ky_dams, x="HEIGHT", ax=ax2
)

ax1.set_xlim(0, 2000)
```
