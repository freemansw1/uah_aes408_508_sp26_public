## Tuples

```python
y = (1, 2, 3)
y
```

```python
y[2]
```

```python
type(y)
```

```python
y[1] = 45
```

```python
y_squared = tuple()
for item in y:
    y_squared.append(item**2)
```

```python
for item in y:
    print(item)
```

```python
x = [(30, -85), (31, -86), (32, -87)]
for item in x:
    print(item)
```
