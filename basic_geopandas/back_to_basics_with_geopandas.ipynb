We're going to go back-to-basics, but we are going to use real geospatial data to hopefully make it a bit more fun. 

```python
import geopandas as gpd
```

We're going to look at the EPA Ecoregions for this: https://www.epa.gov/eco-research/level-iii-and-iv-ecoregions-continental-united-states

```python
epa_files = gpd.read_file("../example_data/us_eco_l3_state_boundaries/us_eco_l3_state_boundaries.shp")
```

```python
epa_files
```

```python
for index, row in epa_files.iterrows():
    print(f"{index}, {row.loc["EPA_REGION"]}")
    break
```

```python
for index, row in epa_files.iterrows():
    print(f"{index}, {row["EPA_REGION"]}")
    break
```

```python
epa_region_counts = dict()
for index, row in epa_files.iterrows():
    print(f"{index}, {row["EPA_REGION"]}")
    epa_region_counts[row["EPA_REGION"]] +=1
    break
```

```python
epa_region_counts = dict()
for index, row in epa_files.iterrows():
    cur_epa_reg = row['EPA_REGION']
    if cur_epa_reg in epa_region_counts:
        epa_region_counts[cur_epa_reg]+=1
    else:
        epa_region_counts[cur_epa_reg] = 1

epa_region_counts
```

```python
total_count = 0
for key in epa_region_counts:
    total_count += epa_region_counts[key]
```
