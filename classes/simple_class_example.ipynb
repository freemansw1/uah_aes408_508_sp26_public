```python
import shapely
```

```python
class House:
    """documentation

    Attributes
    ----------
    address: str
    """

    def __init__(self, address, latitude, longitude):
        self.address = address
        self.coords = shapely.Point(latitude, longitude)
    
    def summary(self):
        print(self.address)

        
```

```python
new_house = House("123 sesame street", 30, -85)
```

```python
new_house.summary()
```
