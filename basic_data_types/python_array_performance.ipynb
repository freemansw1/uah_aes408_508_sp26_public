# Basic Python Array Performance

Recall that this is how we define a basic array

```python
x = [1,2,3,4,5]
```

We can also define an array using an *array comprehension* like this:

```python
x = [i for i in range(1,6)]
x
```

Let's make a long array

```python
x = [i for i in range(13658)]
```

```python
len(x)
```

`%%time` is a special parameter for Jupyter Notebooks that times how long everything in the cell takes to run

```python
%%time
x_squared = list()
for val in x:
    x_squared.append(val**2)
```

Not bad! What happens when we scale this

```python
x = [i for i in range(293071)]
```

```python
%%time
x_squared = list()
for val in x:
    x_squared.append(val**2)
```
