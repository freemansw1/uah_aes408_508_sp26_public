# Basic Python Functions - with types!

```python
from typing import Union, Literal
```

```python
def func(x: Union[int, float]) -> Union[int, float]:
    return x * 2
print(func(5))
```

```python
def func(x: int) -> int:
    return x * 2
print(func(5.78))
```

```python
def func(x: int, op: Literal['+', '-']) -> int:
    if op == '+':
        return x + 2
    elif op == '-':
        return x - 2
print(func(5, '+'))
```
