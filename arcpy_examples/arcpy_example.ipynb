```python
import arcpy
```

```python
in_shapefile_name = r"C:\Users\sfreeman\Downloads\tl_2023_us_county\tl_2023_us_county.shp"
```

```python
aprx = arcpy.mp.ArcGISProject("CURRENT")
cur_map = aprx.listMaps()[0]
```

```python
cur_map
```

```python
lyr = cur_map.addDataFromPath(in_shapefile_name)
```

```python
aprx.listColorRamps()
```

```python
sym = lyr.symbology
sym.updateRenderer("GraduatedColorsRenderer")
sym.renderer.classificationField = "ALAND"
sym.renderer.breakCount = 5
sym.renderer.colorRamp = aprx.listColorRamps("Yellow-Orange-Red (5 Classes)")[0]
lyr.symbology=sym
```

```python
mv = aprx.activeView
mv.camera.setExtent(mv.getLayerExtent(lyr))
```

```python
mv.exportToPNG(r"C:\Users\sfreeman\Downloads\test_counties.png", 
                    width=3000, height=2000, resolution=150)
```

```python
layout = aprx.createLayout(17, 11, "INCH")
layout.name = "County Export"
```

```python
mf = layout.createMapFrame(arcpy.Extent(0.5, 0.5, 16.5, 9.5), cur_map)
```

```python
# add in extent here
```

```python
extent_wgs84 = arcpy.Extent(-125, 24, -66, 50)
```

```python
poly_wgs84 = arcpy.Polygon(arcpy.Array([
    arcpy.Point(extent_wgs84.XMin, extent_wgs84.YMin),
    arcpy.Point(extent_wgs84.XMax, extent_wgs84.YMin),
    arcpy.Point(extent_wgs84.XMax, extent_wgs84.YMax),
    arcpy.Point(extent_wgs84.XMin, extent_wgs84.YMax),]), arcpy.SpatialReference(4326))
```

```python
map_sr = cur_map.spatialReference
```

```python
map_sr
```

```python
poly_projected = poly_wgs84.projectAs(map_sr)
extent_projected = poly_projected.extent
```

```python
north_arrow_cim = arcpy.cim.CreateCIMObjectFromClassName("CIMMarkerNorthArrow", "V3")
north_arrow_cim.mapFrame = mf.name
na_style = aprx.listStyleItems("ArcGIS 2D", "North_Arrow", "ArcGIS North 1")[0]
north_arrow = layout.createMapSurroundElement(arcpy.Extent(15.0, 0.75, 16.0, 1.75), 
                                              "NORTH_ARROW", mf, na_style, "North Arrow" )
```
