# Basic Python Loops

```python
x = [1, 2, 3, 4, 5]
for item in x:
    print(item)
```

```python
x = [545, 215, 3, 48, 0.5]
for item in x:
    print(item)
```

```python
x = [[545,5], [215, 30], [3, 5], [48, 884], [0.5, 9]]
for item in x:
    print(item)
```

```python
x = [[545,5], [215, 30], [3, 5], [48, 884], [0.5, 9]]
for item in x:
    for item_2 in item:
        print(item_2)
```

```python
for i, item in enumerate(x):
    print(f"Index: {i}, Item: {item}")
```

```python
lats = [30, 31, 32, 33, 34, 35]
lons = [-80, -81, -82, -83, -84]

for lat, lon in zip(lats, lons):
    print(lat, lon)
```

## Control Structures

```python
x = [545, 215, 3, 2, 0.5]
for item in x:
    if item == 2:
        print(item)
```

```python
x = [545, 215, 3, 2, 0.5]

for item in x:
    if item == 2:
        continue
    else:
        print(item)
    print("hello!")
```

```python
x = [545, 215, 3, 2, 0.5]

for item in x:
    if item == 2:
        break
    else:
        print(item)
```
